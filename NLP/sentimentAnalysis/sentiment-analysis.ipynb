Basic Sentiment Analysis

Using 'TextBlob' and 'Vader' to check sentiment of the sentence (Positive, Negative or Neutral)

```python
sentence_1 = "i had a great time at the movie it was really funny"
sentence_2 = "i had a great time at the movie but the parking was terrible"
sentence_3 = "i had a great time at the movie but the parking wasn't great"
sentence_4 = "i went to see a movie"
```

TEXTBLOB

the negative values (-1<=x<0) means negative, 0 means neutral, and 0<x<1 positive.

```python
from textblob import TextBlob
```

```python
sentiment_score_1 = TextBlob(sentence_1)
print(sentence_1)
print(sentiment_score_1.polarity)
```

```python
sentiment_score_2 = TextBlob(sentence_2)
print(sentence_2)
print(sentiment_score_2.polarity)
```

```python
sentiment_score_3 = TextBlob(sentence_3)
print(sentence_3)
print(sentiment_score_3.polarity)
```

```python
sentiment_score_4 = TextBlob(sentence_4)
print(sentence_4)
print(sentiment_score_4.polarity)
```

VADER (rangle of values is similar -1 to 1). 

This gives specific and compound sentiment value in the sentence.

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
vader_sentiment = SentimentIntensityAnalyzer()
```

```python
print(sentence_1)
print(vader_sentiment.polarity_scores(sentence_1))
```

```python
print(sentence_2)
print(vader_sentiment.polarity_scores(sentence_2))
```

```python
print(sentence_3)
print(vader_sentiment.polarity_scores(sentence_3))
```

```python
print(sentence_4)
print(vader_sentiment.polarity_scores(sentence_4))
```

USING TRANSFORMERS. Pre-Trained Transformers


```python
#  rule-based analysis is easy to understand but it may not always be accurate.
# transformers are based on deep-learning and are designed to capture how words influence and
# relate to each other even when they appear far apart in sentence.

# simply: Transfomers can understand context.
```

```python
# we are using pipelin(), because it allows quick way to use some pre-trained model without needing to build them from scratch

import transformers
from transformers import pipeline
```

```python
sentiment_pipeline = pipeline("sentiment-analysis") # not provided any sepcific model, will use default one provided.
```

```python
print(sentence_1)
sentiment_pipeline(sentence_1)

#label is what the sentiment is, and the score is how much confidence the model has in that the result is correct.
```

```python
print(sentence_2)
sentiment_pipeline(sentence_2)
```

```python
print(sentence_3)
sentiment_pipeline(sentence_3)
```

```python
print(sentence_4)
sentiment_pipeline(sentence_4)
```

```python
# with a speicific model
specific_model =  pipeline("sentiment-analysis", model="finiteautomata/bertweet-base-sentiment-analysis")
```

```python
print(sentence_1)
specific_model(sentence_1)
```

```python
print(sentence_2)
specific_model(sentence_2)
```

```python
print(sentence_3)
specific_model(sentence_3)
```

```python
print(sentence_4)
specific_model(sentence_4)
```
