```python
customer_reviews = ["sam was a great help to me in the store",
                    "the cashier was very rude to me, I think her name was eleanor",
                    "amazing work from sadeen!", 
                    "sarah was able to help me find the items I needed quickly",
                    "lucy is such a great addition to the team",
                    "great service from sara, she found me what I wanted."]

pattern_to_find = r"sarah?"
sarah_review = [review for review in customer_reviews if re.search(pattern_to_find, review)];
print(sarah_review)
```

```python
# Find all reviews that start with "a"
a_reviews = [review for review in customer_reviews if re.search(r"^a", review)]
print(a_reviews)
```

```python
# Find all reviews that end with "y"
y_reviews = [review for review in customer_reviews if re.search(r"y$", review)]
print(y_reviews)
```

```python
# Find all reviews that contain "need" or "want" followed by "ed"
needwant_reviews = [review for review in customer_reviews if re.search(r"(need|want)ed",review)]
print(needwant_reviews)
```

```python
"""
With brackets [^\w\s] :
[] creates a character class (a set of characters)
^ inside the bracket class means "NOT" (negation)
\w = word characters (letters, digits, underscore)
\s = whitespace
"Match any single character that is NOT a word character and NOT whitespace" --> Matches punctuation

"""

no_punctuation_reviews = [re.sub(r"[^\w\s]", "", review) for review in customer_reviews]
print(no_punctuation_reviews)

#similar to this
""" 
no_punctuation_reviews = []
for review in customer_reviews:
    cleaned = re.sub(r"[^\w\s]", "", review)
    no_punctuation_reviews.append(cleaned)

"""
```

```python
# r"^\w\s", no brackets, I just added it as reference, don't mind if u don't understand it now, only previous code block matters,
# if u run this first before the previous one, u won't get correct output.

# Find all reviews that do not start with a word character followed by a space (i.e., reviews that do not start with punctuation)
# ^ = start of string, \w = word character, \s = space. Word character = a-z, A-Z, 0-9, and _ (underscore). 
no_punctuation_reviews = [re.sub(r"^\w\s", "", review) for review in customer_reviews]
print(no_punctuation_reviews)
```
