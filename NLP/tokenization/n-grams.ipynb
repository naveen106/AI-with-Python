```python
# N-GRAMS: SEQUENCES OF N WORDS IN A SENTENCE. E.G. "I LOVE NLP" CAN BE BROKEN INTO THE FOLLOWING N-GRAMS:
# UNIGRAMS: ["I", "LOVE", "NLP"]
# BIGRAMS: ["I LOVE", "LOVE NLP"]
# TRIGRAMS: ["I LOVE NLP"]

import nltk
import pandas as pd # pandas is the library used for working with structured data. Dataframe is core feature which works like spreadsheet inside Python
import matplotlib.pyplot as plt # matplotlib is a library used for data visualization. e.g. chart, graph, etc.

tokens =    ['the', 'rise', 'of', 'artificial', 'intelligence', 'has', 'led', 'significant', 'advancements', 'in', 'natural', 
            'language', 'processing', 'and', 'machine', 'learning', 'and', 'other',  'fields', 'machine', 'learning', 'algorithms', 'have', 'been', 'developed', 'to', 'analyze', 'and', 'understand',    
            'natural', 'language', 'data', 'and', 'generate', 'human-like', 'text', 'and', 'other', 'outputs', 'these', 'advancements', 'have', 'led', 'to', 'the', 'development', 'of', 'chatbots', 'virtual', 'assistants', 'and',
            'other', 'applications', 'that', 'can', 'interact', 'with', 'humans', 'in', 'a', 'natural', 'and', 'intuitive', 'way',  'overall', 'the', 'rise', 'of', 'artificial', 'intelligence', 'has', 'had', 'a', 'profound', 'impact', 'on', 
            'the', 'field', 'of', 'natural', 'language', 'processing', 'and', 'machine', 'learning', 'and', 'has', 'led', 'to', 'the', 'development', 'of', 'many', 'innovative', 'applications', 'that', 'are', 'transforming', 
            'the', 'way', 'we', 'interact', 'with', 'technology', 'and', 'each', 'other']

unigrams = pd.Series(nltk.ngrams(tokens, 1)).value_counts()
print(unigrams[:10]) # print the 10 most common unigrams

```

```python
unigrams[:10].sort_values().plot.barh(color='skyblue', width=0.6, figsize=(7,4))
plt.title("Most Frequntly Appearing Unigrams")
```

```python
bigrams = pd.Series(nltk.ngrams(tokens, 2)).value_counts()
print(bigrams[:10]);
```

```python
trigrams = pd.Series(nltk.ngrams(tokens,3)).value_counts()
print(trigrams[:10])
```
