```python
import nltk
#   STEMMING: REDUCING WORDS TO THEIR ROOT FORM E.G. "RUNNING" TO "RUN"
from nltk.stem import PorterStemmer

porter_stemmer = PorterStemmer()
connect_tokens = ["connected", "connecting", "connection", "connections", "connectivity", "connects", "connector"]

for token in connect_tokens:
    print(token, " : ", porter_stemmer.stem(token))
```

```python
learn_tokens = ["learned", "learning", "learner", "learns", "learnt", "learners"]

for token in learn_tokens:
    print(token," : ", porter_stemmer.stem(token))
```

```python
like_tokens = ["like", "better","worse"]

# here worse is stemmed to "wors" which is not a valid word. This is a limitation of stemming. It can sometimes produce non-words.
for token in like_tokens:
    print(token, " : ",  porter_stemmer.stem(token))
```

```python
#LEMMATIZATION: REDUCING WORDS TO THEIR BASE FORM E.G. "RUNNING" TO "RUN"

# wordnet is a lexical database for the English language that provides a large collection of words 
# and their relationships, including synonyms, antonyms, and hierarchical relationships. 
nltk.download('wordnet') 
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer =  WordNetLemmatizer()

for token in connect_tokens:
    print(token, " : ",lemmatizer.lemmatize(token))
```

```python
for token in learn_tokens:
    print(token, " : ", lemmatizer.lemmatize(token))
```

```python
for token in like_tokens:
    print(token," : ", lemmatizer.lemmatize(token))

# lemmatization is more accurate than stemming as it considers the context and the meaning of the word.
```
