```python
import nltk #natural language toolkit
from nltk.tokenize import word_tokenize #tokenize the words
from nltk.stem import PorterStemmer, WordNetLemmatizer # stem the words to basic form or use lemmatize to look up their meaning
from nltk.corpus import stopwords #get stopwords to get meaningful words only.
import pandas #to work with structured data
import matplotlib.pyplot #to plot data in graph, charts, etc
import re # for regex
```

```python
data = pandas.read_csv("tripadvisor-hotel-reviews.csv") # dataset of hotel reviews we are working with
data.info()

# in the output here, "Non-Null Count" is one attribute/column, 
# they are not seperate, so  "109 non-null" means that many non-null rows are present
```

```python
data.head()
```

```python
data['Review'][0] #checking one review in dataset
```

```python
#adding the same reviews in another column with all words lowercase completely
data['review_lowercase'] = data['Review'].str.lower()
data.head()
```

```python
en_stopwords = stopwords.words('english');
en_stopwords.remove('not')

data['review_no_stopwords']  = data['review_lowercase'].apply(lambda review : ' '.join(word for word in review.split() if word not in en_stopwords))
data.head()
```

```python
data['review_no_stopwords'][0]
```

```python
data['review_no_punct_no_stopwords'] = data.apply(lambda row : re.sub(r"[*]", "star", row['review_no_stopwords']), axis=1)
```

```python
data.head()
```

```python
data['review_no_punct_no_stopwords'] = data.apply(lambda row : re.sub(r"[^\w\s]", "", row["review_no_punct_no_stopwords"]), axis=1)
```

```python
data.head()
```

```python
data['tokenized'] = data.apply(lambda row : word_tokenize(row['review_no_punct_no_stopwords']), axis=1)
data['tokenized'][0]
```

```python
ps = PorterStemmer()
data['stemmed'] = data['tokenized'].apply(lambda tokens : [ps.stem(token) for token in tokens])
data.head()
```

```python
lemmatizer = WordNetLemmatizer()
data['lemmatized'] = data['tokenized'].apply(lambda tokens : [lemmatizer.lemmatize(token) for token in tokens])

data.head()
```

```python
data['lemmatized'][0]
```

```python
tokens_clean = sum(data['lemmatized'],[]) #adds all tokens in one list, starting from empty list.
```

```python
unigrams = pandas.Series(nltk.ngrams(tokens_clean, 1)).value_counts()
```

```python
print(unigrams)
```

```python
bigrams = pandas.Series(nltk.ngrams(tokens_clean, 2)).value_counts()
```
